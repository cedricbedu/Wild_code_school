```python
# J'installe mes bibliothèques
from curl_cffi import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Je définis l'url que je veux scraper
url = 'https://www.chucknorrisfacts.fr/facts/top/1'

# Je récupere les infos grace au .get()
response = requests.get(url, impersonate="chrome")

# Falcutatif : la gestion d'erreur
if response.status_code == 200:
    html = BeautifulSoup(response.text)
    print("Scraping effectué avec succès, le response code est 200")
else:
    print(f"Erreur lors du scraping : {response.status_code}")
```

```python
# J'affiche le nombre de divs
divs = html.find_all('div', class_='card')
print(f"Nombre de divs trouvés : {len(divs)}")
```

```python
# J'affiche le texte de la 8eme div
divs[8].text
```

```python
# Huitième blague (index 8) = donc blague 9 (la 8 etant vide)

blague_8 = divs[8]

# blague

blague = blague_8.find("p", class_="card-text").get_text()
print(f"Voici la blague : {blague}")

# note de la blague

note = blague_8.find("div", class_="row").find("span", id="moyenne_100410").get_text()
print(f"Note de la blague : {note}")
```

```python
# Je récupere les notes
for i in divs :
    row = i.find("div", class_="row")
    if row :
        note = row.find("span").get_text()
        print(note)
```

```python

# Je récupère les blagues
for i in divs :
    row = i.find("div", class_="row")
    if row :
        blague = i.find("p", class_="card-text").get_text()
        print(blague)
```

```python
# Je crée un dictionnaire vide 
liste_blague = []

# Je remplis mon dictionnaire avec les blagues et les notes
for i in divs :
    row = i.find("div", class_="row")   # Je crée une variable row qui permet de vérifier si il y a bien une valeur dans la div
    # Si il y a une valeur dans la div, alors je remplis mon dictionnaire avec la blague et la note
    if row : 
        dico = {"Blague" : i.find("p", class_="card-text").get_text(), "Note" : row.find("span").get_text()}
        liste_blague.append(dico)
display(liste_blague)
```

```python
# Je transforme mon dictionnaire en dataframe

df = pd.DataFrame(liste_blague)
df
```
