```python
# Import des packages
from curl_cffi import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Ci dessous l'url permettant d'avoir les offres d'emploi sur linkedin :

url = "https://www.linkedin.com/jobs-guest/jobs/api/seeMoreJobPostings/search?keywords=data&location=france&geoId=&trk=public_jobs_jobs-search-bar_search-submit&start=0"
```

1. Identifier les parametres important de l'url

Les paramètres importants de l'url sont après le "search?", on y voit la situation géographique (France), le trk montre que ca concerne les "public_jobs" et le start est a 0. Le paramètre keywords correspond à data, les jobs recherchés sont donc dans la data, le geoID n'est pas mentionné, 

2. Requeter l'url , est ce que votre requete est acceptée ? Comment le savoir ?

```python
html = requests.get(url)
print(f"Ma réponse est : {html.status_code}")
```

La réponse est 200 donc ma requête est acceptée

3. Parser le contenu recupéré avec bs4 :

```python
soup = BeautifulSoup(html.text, 'html.parser')
print(soup)
```

4. Récuperer le titre de l'annonce, le nom de l'entreprise, localisation , lien du détail de l'offre et la date de l'offre :


```python
# J'affiche le nombre de divs
divs = soup.find_all('div', class_='base-card')
print(f"Nombre de divs trouvés : {len(divs)}")
```

```python
# J'affiche le texte le la première div

divs[0].text
```

```python
# Je crée une variable pour stocker mon premier div

divn1 = divs[0]
```

```python
# Je cherche a trouver le titre de l'offre 

job = divn1.find('h3', class_='base-search-card__title').get_text().strip()
print(f"Le premier job est : {job}")
```

```python
# Maintenant je veux boucler pour recuperer tous les jobs

# Je récupère les jobs
for i in divs :
    job = i.find('h3', class_='base-search-card__title').get_text().strip()
    print(job)

# Je l'affiche en compréhension de liste et je le stocke dans une variable
boucle_job = [i.find('h3', class_='base-search-card__title').get_text().strip() for i in divs]
```

```python
# Maintenant je fais pareil pour le nom de l'entreprise 

# Je cherche a trouver le nom de l'entreprise

entreprise = divn1.find('h4', class_='base-search-card__subtitle').get_text().strip()
print(f"La premiere entreprise est : {entreprise}")
```

```python
# Maintenant je veux boucler pour recuperer toutes les entreprises

# Je récupère les entreprise
for i in divs :
    entreprise = i.find('h4', class_='base-search-card__subtitle').get_text().strip()
    print(entreprise)

# Je l'affiche en compréhension de liste et je le stocke dans une variable
boucle_entreprise = [i.find('h4', class_='base-search-card__subtitle').get_text().strip() for i in divs]
```

```python
# Maintenant je fais pareil pour le lien de l'offre 

# Je cherche a trouver le lien de l'offre

offre = divn1.find('a')['href']
print(f"Le lien de la premiere offre est : {offre}")
```

```python
# Maintenant je veux boucler pour recuperer touts les liens

# Je récupère les liens
for i in divs :
    lien = i.find('a')['href']
    print(lien)

# Je l'affiche en compréhension de liste et je le stocke dans une variable
boucle_lien = [i.find('a')['href'] for i in divs]
```

```python
# Maintenant je fais pareil pour la date de l'offre 

# Je cherche a trouver la date de l'offre

date = divn1.find('time')['datetime']
print(f"La date de la première offre est : {date}")
```

```python
# Maintenant je veux boucler pour recuperer toutes les dates

# Je récupère les dates
for i in divs :
    date = i.find('time')['datetime']
    print(date)

# Je l'affiche en compréhension de liste et je le stocke dans une variable
boucle_date = [i.find('time')['datetime'] for i in divs]
```

5. Stocker les datas récupérées dans un dataframe:


```python
df_linkedin = pd.DataFrame({"jobs" : boucle_job,
              "entreprise" : boucle_entreprise,
              "lien" : boucle_lien,
              "date" : boucle_date})

df_linkedin
```

```python
# Je crée une fonction pour créer des liens avec l'id

def url_id(lien):
    id = re.search(r"\d{10}", lien).group()
    lien = f"https://www.linkedin.com/jobs-guest/jobs/api/jobPosting/{id}"
    return lien

df_linkedin["info_id"] = df_linkedin['lien'].apply(url_id)

df_linkedin.head()
```
